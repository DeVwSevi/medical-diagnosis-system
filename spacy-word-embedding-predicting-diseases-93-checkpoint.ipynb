# Importing Basic Libraries

```python
import pandas as pd
import numpy as np
```

# Reading csv

```python
df = pd.read_csv("data/Symptom2Disease.csv")
df.head()
```

# Mapping numbers to 24 unique labels

```python
df.label.unique()
```

```python
df['label_num']= df.label.map(dict(zip(pd.Series(df['label'].unique()),pd.Series([i for i in range(24)]))))
                              
df.head()
```

```python
df.label_num.nunique()
```

# Importing SPACY

## For Preprocessing of text

```python
!pip install spacy
```

```python
import spacy

nlp = spacy.load("en_core_web_lg") 

def preprocess(text):
    list =[]
    for token in nlp(text):
        if token.is_space or token.is_punct:
            continue
        list.append(token.lemma_)
    return ' '.join(list)
```

```python
df['preprocess'] = df['text'].apply(preprocess)
```

## For Vectorizing of text

```python
df['vector'] = df.preprocess.apply(lambda text: nlp(text).vector)
df.head()
```

# Splitting Data into Train and Test

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df['vector'].values,df['label_num'],test_size= 0.30 , random_state = 23 , stratify= df['label_num'])
```

# Changing Train and Test data into 2d array

```python
X_train_2d = np.stack(X_train)
X_test_2d = np.stack(X_test)
```

# Scaling of Data

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

X_train_scaled = scaler.fit_transform(X_train_2d)
X_test_scaled = scaler.transform(X_test_2d)
```

# Importing SVC for model fitting

```python
from sklearn.svm import SVC

model = SVC(C=1, kernel='poly', degree=3, gamma="scale")
    
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)
```

# Classification Report : 93% accuracy. Hurray !

```python
from sklearn.metrics import classification_report

print('\n\n\n Classification Report   :\n\n\n ' , classification_report(y_test,y_pred))
```

# Heatmap of Confusion Matrix

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test , y_pred)
```

```python
import matplotlib.pyplot as plt
import seaborn as sn

font = {'family':'fantasy','color':'blue','size':20}
plt.figure(figsize=(9,9), facecolor= 'yellow', edgecolor='red')
sn.heatmap(cm , annot= True, fmt="d")
plt.xlabel('Prediction', fontdict= {'family':'fantasy','color':'red','size':15})
plt.ylabel('Truth', fontdict= {'family':'fantasy','color':'red','size':15})
plt.title('Confusion Matrix',fontdict= {'family':'fantasy','color':'blue','size':20})
plt.show()
```
